```python
#FOR LOOP
attendance = [85,72,91,65,78]
for percentage in attendance:
  if percentage >= 75:
    print("Eligible for exam")
  else:
    print("Not eligible for exam")
```

```python
#WHILE LOOP
correct_password="python123"
attempts=0

while attempts<3:
  password=input("Enter password:")
  attempts += 1

  if password==correct_password:
    print("login successful!")
    break
  else:
    print("Incorrect password.Try again.")

if attempts==3 and password != correct_password:
  print("Account locked!")
```

```python
#function to calculate the avg marks:
def calculate_avg_marks(sub1_mark,sub2_mark,sub3_mark):
  total_marks=sub1_mark+sub2_mark+sub3_mark
  avg_marks=total_marks//len(marks)
  return avg_marks

#function call
marks=[100,80,80,90]
avg_marks=calculate_avg_marks(100,50,80)
print("Average marks:",avg_marks)
```

# SDP11

```python
#exception handling
try:#code that my cause error
  result=10/0
except ZeroDivisionError:#error type
  print("Error:Division by zero")
finally:#executes cleanup code always
  print("finally block")
```

```python
#ex
try:
  num1 = int(input("Enter 1 number: "))
  num2 = int(input("Enter 2 number: "))
  result = num1 / num2
  print("Result:", result)
except ZeroDivisionError:
  print("Error:Division by zero")
except ValueError:
  print("Error:Invalid input.Please enter a valid number.")
finally:
  print("finally block")
```
