# SDP5

```python
marks = int(input("Enter your marks:"))
attendance = int(input("Enter your attendance percentage:"))
```

```python
if marks >= 50 and attendance >= 75:
  print("Eligible for exam")
elif marks < 50:
  print("Not eligible for exam because marks are less than 50%")
else:
  print("Not eligible for exam because attendance is less than 75%")
```

# SDP6


```python
#FOR LOOP
attendance = [85,72,91,65,78]
for percentage in attendance:
  if percentage >= 75:
    print("Eligible for exam")
  else:
    print("Not eligible for exam")
```

```python
#WHILE LOOP
correct_password="python123"
attempts=0

while attempts<3:
  password=input("Enter password:")
  attempts += 1

  if password==correct_password:
    print("login successful!")
    break
  else:
    print("Incorrect password.Try again.")

if attempts==3 and password != correct_password:
  print("Account locked!")
```
