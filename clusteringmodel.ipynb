```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn_extra.cluster import KMedoids
from sklearn.metrics import silhouette_score,davies_bouldin_score
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
x=np.random.randn(50,2)
```

```python
kmeans=KMeans(n_clusters=3,random_state=42)
kmeans.fit(x)
centers=kmeans.cluster_centers_
Klabels=kmeans.labels_
kmeans_inertia=kmeans.inertia_
```

```python
kmeans_silhouette=silhouette_score(x,Klabels)
kmeans_db=davies_bouldin_score(x,Klabels)
```

```python
kmedoids=KMedoids(n_clusters=3,random_state=42)
kmedoids.fit(x)
medoids=kmedoids.cluster_centers_
KMlabels=kmedoids.labels_
```

```python
kmedoids_silhouette=silhouette_score(x,KMlabels)
kmedoids_db=davies_bouldin_score(x,KMlabels)
```

```python
print("K-Means:")
print(f"Silhouette Score:{kmeans_silhouette:.4f}")
print(f"Davies-Bouldin Index:{kmeans_db:.4f}")
print(f" Inertia(within-cluster sum of squares):{kmeans_inertia:.4f}\n")
```

```python
print("K-Medoids:")
print(f"Silhouette Score:{kmedoids_silhouette:.4f}")
print(f"Davies-Bouldin Index:{kmedoids_db:.4f}")
```

```python
print("\nInterpretation:")
print("Higher silhouette score -> well separated clusters")
print("Lower Davies-Bouldin index -> better clustering")
print("For k-means, lower inertia -> more compact clusters")
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.scatter(x[:,0],x[:,1],c=Klabels,cmap='viridis',marker='o')
plt.scatter(centers[:,0],centers[:,1],c='red',s=100, marker='x', label='Centroids')
plt.title('K-Means Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()

plt.subplot(1,2,2)
plt.scatter(x[:,0],x[:,1],c=KMlabels,cmap='viridis',marker='o')
plt.scatter(medoids[:,0],medoids[:,1],c='red',s=100, marker='D', label='Medoids')
plt.title('K-Medoids Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()

plt.tight_layout()
plt.show()
```
