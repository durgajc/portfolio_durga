```python
import pandas as pd
import numpy as np
data=pd.DataFrame({
    'age':[25,np.nan,30,45,np.nan],
    'salary':[30000,40000,np.nan,50000,70000],
    'city':['New York','Los Angeles','New York','San Francisco',np.nan],
    'target':[1,0,1,0,1]
})
```

```python
print("og data", data)
```

```python
print("missing values:", data.isnull().sum())
```

```python
print("missing value percentage:", data.isnull().mean()*100)
```

```python
data=data.dropna()
print("af remv rows:", data)
```

```python
data=data.dropna(axis=1)
print("af remv col:", data)
```

```python
from sklearn.impute import SimpleImputer
```

```python
num_features=['age','salary']
imputer_num=SimpleImputer(strategy='mean')
data[num_features]=imputer_num.fit_transform(data[num_features])
```

```python
cat_features=['city']
imputer_cat=SimpleImputer(strategy='most_frequent')
data[cat_features]=imputer_cat.fit_transform(data[cat_features])
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
cat_features=['city']
label_encoder={}
for col in cat_features:
  le=LabelEncoder()
  data[col]=le.fit_transform(data[col])
  label_encoder[col]=le
```

```python
scaler = StandardScaler()
num_features = ['age', 'salary']
data[num_features] = scaler.fit_transform(data[num_features])
data.to_csv('clean dataset', index=False)
```
